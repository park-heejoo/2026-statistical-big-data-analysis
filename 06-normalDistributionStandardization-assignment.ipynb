```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
np.random.seed(42)
```

**실습(1)**  
• 정규분포의 표본 평균, 표본분산, 이론적 평균, 이론적 분산 출력하기  
• 정규분포 시각화  
&emsp;-데이터의 분포는 히스토그램, 확률밀도함수는 선그래프  
&emsp;-그래프의 x축 범위는 np.linspace(mu-4*sigma, mu+4*sigma, 1000) 사용  
&emsp;-확률밀도함수는 norm.pdf() 함수 사용

```python
mu=5.0 #평균
sigma=0.051 #표준편차
n=10000 #표본 개수

data=np.random.normal(loc=mu, scale=sigma, size=n)
```

```python
#정규분포의 통계치 계산
sample_mean = np.mean(data)
sample_var = np.var(data, ddof=1) #표본 분산
theoretical_mean = mu
theoretical_var = sigma**2 #이론적 분산

print(f"--- 통계 결과 ---")
print(f"표본 평균 (Sample Mean): {sample_mean:.6f}")
print(f"표본 분산 (Sample Variance): {sample_var:.6f}")
print(f"이론적 평균 (Theoretical Mean): {theoretical_mean:.6f}")
print(f"이론적 분산 (Theoretical Variance): {theoretical_var:.6f}")
print("-" * 20)

#정규분포 시각화
x = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)

plt.figure(figsize=(10, 6))

#히스토그램
plt.hist(data, bins=50, density=True, alpha=0.6, color='skyblue', label='Sample Data (Histogram)')

# 확률밀도함수(PDF)
pdf = norm.pdf(x, mu, sigma)
plt.plot(x, pdf, 'r-', lw=2, label='Theoretical PDF (Line Graph)')

plt.title(f'Normal Distribution (n={n})')
plt.xlabel('X')
plt.ylabel('Density')
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
```

**실습(2)**  
• 실습1의 데이터를 z스코어로 변환하여 표준정규분포로 만들기  
• 표준정규분포의 표본평균, 표본분산, 이론적 평균, 이론적 분산 출력하기  
• 표준정규분포 시각화  
&emsp;-데이터의 분포는 히스토그램, 확률밀도함수는 선그래프  
&emsp;-그래프의 x축 범위는 np.linspace(mu-4*sigma, mu+4*sigma, 1000) 사용  
&emsp;-확률밀도함수는 norm.pdf() 함수 사용

```python
z_data=(data-mu)/sigma

#표준정규분포의 통계치 계산
s_mean=np.mean(z_data)
s_var=np.var(z_data, ddof=1)
t_mean=0
t_var=1

print(f"--- z-score 변환 후 표준정규분포 데이터 ---")
print(f"표본 평균 (Sample Mean): {s_mean}")
print(f"표본 분산 (Sample Variance): {s_var}")
print(f"이론적 평균 (Theoretical Mean): {t_mean}")
print(f"이론적 분산 (Theoretical Variance): {t_var}")
print("-" * 30)

#표준정규분포 시각화
mu_std, sigma_std = 0, 1
x = np.linspace(mu_std - 4*sigma_std, mu_std + 4*sigma_std, 1000)

plt.figure(figsize=(10, 6))

#히스토그램
plt.hist(z_data, bins=50, density=True, alpha=0.5, color='gold', label='Z-score Data (Histogram)')

# 확률밀도함수(PDF)
pdf = norm.pdf(x, mu_std, sigma_std)
plt.plot(x, pdf, 'b-', lw=2, label='Standard Normal PDF (Line)')

plt.title('Standard Normal Distribution (Z-score Transformation)')
plt.xlabel('Z-value')
plt.ylabel('Density')
plt.axvline(0, color='black', linestyle='--', alpha=0.5) # 평균(0) 지점 표시
plt.legend()
plt.grid(True, alpha=0.3)

plt.show()
```
