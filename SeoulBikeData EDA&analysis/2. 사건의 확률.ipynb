```python
import pandas as pd
import numpy as np

df = pd.read_csv('SeoulBikeData.csv', encoding='unicode_escape')
df.columns = ['Date', 'Rented Bike Count', 'Hour', 'Temperature', 'Humidity', 
              'Wind_speed', 'Visibility', 'Dew_point_temp', 'Solar_Radiation', 
              'Rainfall', 'Snowfall', 'Seasons', 'Holiday', 'Functioning Day']
```

```python
# 전체 대여 건수
total = df['Rented Bike Count'].sum()
df['Date'] = pd.to_datetime(df['Date'], dayfirst=True)

# 사건 A: 출퇴근 시간 대여량 (평일 08시, 18시)
event_a = df[df['Hour'].isin([8, 18]) & (df['Date'].dt.dayofweek < 5) & (df['Holiday'] == 'No Holiday')]
p_a = event_a['Rented Bike Count'].sum() / total

# 사건 B: 주중 대여량
event_b = df[df['Date'].dt.dayofweek < 5]
p_b = event_b['Rented Bike Count'].sum() / total

# 사건 C: 서비스 정상 운영 확률
event_c = len(df[df['Functioning Day'] == 'Yes'])
p_c = event_c / len(df)

#사건 D: 대중교통 단절(심야 시간: 0,1,2,3,4시)
event_d=df[df['Hour'].isin([0, 1, 2, 3, 4])]
p_d= event_d['Rented Bike Count'].sum() / total

print(f"P(A) 출퇴근 시간에 대여될 확률: {p_a:.4f} ({p_a*100:.1f}%)" )
print(f"P(B) 주중에 대여될 확률: {p_b:.4f} ({p_b*100:.1f}%)")
print(f"P(C) 서비스가 정상적으로 제공될 확률: {p_c:.4f} ({p_c*100:.1f}%)")
print(f"P(D) 심야 시간에 대여될 확률: {p_d:.4f} ({p_d*100:.1f}%)")
```
