```python
import seaborn as sns
import matplotlib.pyplot as plt

tips=sns.load_dataset('tips')
```

**임의의 수치형 데이터 2개를 선택후, 히스토그램 시각화**  
Seaborn 라이브러리의 histplot() 함수를 사용

```python
sns.histplot(data=tips, x='total_bill', bins=20, kde=True)

plt.title('Distribution of Total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Count')
plt.show()
```

```python
sns.histplot(data=tips, x='tip', bins=20, kde=True)

plt.title('Distribution of Total Bill')
plt.xlabel('Tip')
plt.ylabel('Count')
plt.show()
```

**수치형 데이터들의 상관계수를 구한 후, 히트맵 시각화**  
Seaborn 라이브러리의 heatmap() 함수를 사용

```python
corr = tips.corr(numeric_only=True)

#히트맵 시각화
plt.figure(figsize=(8, 6))
sns.heatmap(corr, 
            annot=True,      # 각 셀에 숫자를 표시
            fmt=".2f",       # 소수점 둘째 자리까지 표시
            cmap='coolwarm', # 색상 맵 (파랑-하양-빨강)
            vmin=-1,  # 셀 사이의 간격
            vmax=1)     # 각 셀을 정사각형으로 설정

plt.title('Correlation Heatmap')
plt.show()
```

**임의의 수치형 데이터 2개를 선택 후, 산점도 시각화**  
Seaborn 라이브러리의 scatterplot() 함수를 사용

```python
sns.scatterplot(
    data=tips,
    x='size',
    y='tip'
)

plt.title('Size vs Tip')
plt.xlabel('Size')
plt.ylabel('Tip')
plt.show()
```

**임의의 수치형 데이터 1개, 범주형 데이터 1개 선택 후, 박스플롯 시각화**  
Seaborn 라이브러리의 boxplot() 함수를 사용

```python
#박스플롯 시각화
plt.figure(figsize=(10, 6))
sns.boxplot(
    data=tips, 
    x='day',           # 범주형 데이터 (요일)
    y='total_bill',     # 수치형 데이터 (전체 요금)
)

#제목 및 라벨 추가
plt.title('Distribution of Total Bill by Day')
plt.xlabel('Day', fontsize=12)
plt.ylabel('Total Bill ($)', fontsize=12)

plt.show()
```

**임의의 범주형 데이터 2개 선택 후, 그룹별 막대그래프 시각화**

```python
import pandas as pd

day_time_ratio=pd.crosstab(
    tips['day'],
    tips['sex']
)

day_time_ratio.plot(
    kind='bar',
    stacked=False
)

plt.title('Number of Visits by Day and Gender')
plt.xlabel('Day')
plt.ylabel('Count')
plt.legend(title='Gender')
plt.show()
```
