```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import randint, bernoulli, binom

np.random.seed(42) #랜덤 결과가 항상 똑같이 나오도록 고정(42는 의미없는 숫자)
```

이산균등분포

```python
low, high=1,7 #추출할 수의 범위 지정(1~6)
n_samples=1000 #뽑을 샘플 개수(1000개)

uniform_samples=np.random.randint(low, high, size=n_samples) #설정한 범위(1~6)내에서 정수 1000개를 무작위로 생성하여 배열에 저장
sample_mean=np.mean(uniform_samples) #생성된 1000개 샘플의 실제 평균(표본평균)을 계산
sample_var=np.var(uniform_samples) #생성된 1000개 샘플의 실제 분산(표본분산)을 계산

a,b=low, high-1 #공식에 대입하기 위해 시작값 a=1, 끝값 b=6으로 변수 지정
theoretical_mean=(a+b)/2 #이산균등분표의 이론적 기댓값 공식
theoretical_var=((b-a+1)**2-1)/12 #이산균등분포의 이론적 분산 공식

print(f'표본평균: {sample_mean:.4f}')
print(f'표본분산: {sample_var:.4f}')
print(f'이론적 기댓값 E(X): {theoretical_mean:.4f}')
print(f'이론적 분산 Var(X): {theoretical_var:.4f}')
```

이산균등분포 시각화

```python
x_uniform=np.arange(low, high) #x축에 표시할 정수 값(1~6)
pmf_uniform=randint.pmf(x_uniform, low, high) #각 숫자가 나올 이론적인 확률 계산

plt.figure(figsize=(8,4)) #창 크기
plt.hist(uniform_samples, bins=np.arange(low-0.5, high+0.5,1), density=True, alpha=0.6, edgecolor='black')
plt.plot(x_uniform, pmf_uniform, 'ro', label='Theoretical PMF') #계산해둔 이론적 확률을 빨간색 동그라미(ro) 점으로 표시
plt.xticks(x_uniform) #x축에 1~6에 맞춰 눈금 설정
plt.title('Discrete Uniform Distribution')
plt.xlabel('x')
plt.ylabel('Probability')
plt.legend() #범례(Theoretical PMF 등)를 화면에 표시
plt.show()
```
