```python
import seaborn as sns

df=sns.load_dataset('penguins')
```

```python
print(df.head())
```

```python
print('Before ->', df.shape)
df=df.dropna()
print('After ->', df.shape)
```

```python
print(f'표본공간(Penguins 데이터셋에 포함된 전체 펭귄의 집합): {len(df)}')
```

```python
A=df['species']=='Adelie'
B=df['flipper_length_mm']>=200

print(f'사건 A(Adelie 종): {A.sum()}개체')
print(f'사건 B(지느러미 길이 200mm 이상): {B.sum()}개체')
```

```python
#합사건
A_or_B=A|B
print(f'A와 B의 합사건(Adelie 종이거나 지느러미 길이 200mm 이상): {A_or_B.sum()}개체')

#곱사건
A_and_B=A&B
print(f'A와 B의 곱사건(Adelie 종이면서 지느러미 길이 200mm 이상): {A_and_B.sum()}개체')
```

```python
#여사건
not_A=~A
print(f'A의 여사건(Adelie 종이 아님): {not_A.sum()}개체')

#차사건
A_minus_B=A&~B
print(f'A와 B의 차사건(Adelie 종이면서 지느러미 길이가 200mm 미만): {A_minus_B.sum()}개체')
```

```python
#사건 A의 확률
print(f'사건 A의 확률: {A.sum()/len(df):.4f}')
print(f'사건 B의 확률: {B.sum()/len(df):.4f}')
print(f'A와 B의 합사건의 확률: {A_or_B.sum()/len(df):.4f}')
print(f'A와 B의 곱사건의 확률: {A_or_B.sum()/len(df):.4f}')
print(f'A의 여사건 확률: {not_A.sum()/len(df):.4f}')
print(f'A와 B의 차사건의 확률: {A_minus_B.sum()/len(df):.4f}')
```
