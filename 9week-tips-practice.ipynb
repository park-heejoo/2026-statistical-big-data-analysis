**Tips 데이터셋**  
식당에서 손님들이 지불한 총액과 팁 금액, 손님 특성, 방문 요일·시간 등을 기록한 소규모 팁 데이터

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips=sns.load_dataset('tips')
```

**두 수치형 데이터 산점도**

```python
sns.scatterplot(
    data=tips,
    x='total_bill',
    y='tip'
)

plt.title('Total Bill vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```

**두 범주형 데이터 그룹 막대그래프**

```python
import pandas as pd

day_time_ratio=pd.crosstab(
    tips['day'],
    tips['time']
)

day_time_ratio.plot(
    kind='bar',
    stacked=False
)

plt.title('Conditional Distribution of Time by Day')
plt.xlabel('Day')
plt.ylabel('Proportion')
plt.legend(title='Time')
plt.show()
```
