```python
#데이터셋 불러오기
from sklearn.datasets import fetch_openml
X,y=fetch_openml('titanic', version=1, as_frame=True, return_X_y=True, parser='auto')
```

```python
#사용할 데이터 선택
import pandas as pd

df=pd.concat([X,y],axis=1)
df=df[['pclass','sex','age','sibsp','parch','fare','embarked','survived']]
```

```python
#데이터 전처리
df['survived']=df['survived'].astype(int)
df['pclass']=df['pclass'].astype(int)
df=df.dropna().reset_index(drop=True)
```

```python
#조건부 확률
#여성 승객이 생존할 확률 P(생존|여성)
p_female=(df['sex']=='female').sum()/len(df)
p_survive_and_female=((df['sex']=='female')&(df['survived']==1)).sum()/len(df)
p_survive_given_female=p_survive_and_female/p_female

print(f'P(여성)={p_female:.4f}')
print(f'P(생존∩여성)={p_survive_and_female:.4f}')
print(f'P(생존|여성)={p_survive_and_female:.4f}/{p_female:.4f}={p_survive_given_female*100:.2f}%')
```

```python
#독립 여부 판단
#승객이 여싱인 사건A, 승객이 생존한 사건B
event_a=(df['sex']=='female')
event_b=(df['survived']==1)

p_a=event_a.sum()/len(df)
p_b=event_b.sum()/len(df)
p_a_and_b=(event_a&event_b).sum()/len(df)
if p_a_and_b==(p_a*p_b):
    print('두 사건은 독립 (Independent)')
else:
    print('두 사건은 종속 (Dependent)')
```

```python
#전확률
#성별을 분할로 전체 사망률 계산
male=df[df['sex']=='male']
female=df[df['sex']=='female']

p_b1=len(male)/len(df)
p_b2=len(female)/len(df)

p_death_b1=1-(male['survived'].sum()/len(male))
p_death_b2=1-(female['survived'].sum()/len(female))

p_death=p_death_b1*p_b1+p_death_b2*p_b2

print(f'P(사망|남성) x P(남성)={p_death_b1:.4f}x{p_b1:.4f}={p_death_b1*p_b1:.4f}')
print(f'P(사망|여성) x P(여성)={p_death_b2:.4f}x{p_b2:.4f}={p_death_b2*p_b2:.4f}')
print(f'P(사망)={p_death:.4f} / 직접 계산:{1-df["survived"].sum()/len(df):.4f}')
```

```python
#베이즈 정리
#생존했다는 사실을 알 때, 그 사람이 몇 등석(pclass)일 확률

groups=sorted(df['pclass'].unique())
p_evidence=(df['survived']==1).sum()/len(df)

for g in groups:
    grp=df[df['pclass']==g]
    p_bi=len(grp)/len(df)
    p_a_bi=(grp['survived']==1).sum()/len(grp)
    p_bi_a=(p_a_bi*p_bi)/p_evidence
    print(f'pclass가 {g}일 때')
    print(f'-사전확률 P(b_i)={p_bi:.4f}')
    print(f'-사후확률 P(b_i|a)={p_bi_a:.4f}')
```
