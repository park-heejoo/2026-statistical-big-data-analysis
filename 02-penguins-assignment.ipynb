```python
import seaborn as sns

df=sns.load_dataset('penguins')
```

```python
#실습 (1)
df=df.dropna()
C=df['island']=='Biscoe'
D=df['sex']=='Male'
E=df['body_mass_g']>=4000
F=df['bill_length_mm']>df['bill_length_mm'].mean()
```

```python
print(f'사건 C(Biscoe 섬에 서식): {C.sum()}개체 / 확률: {C.sum()/len(df):.4f}') 
print(f'사건 D(수컷): {D.sum()}개체 / 확률: {D.sum()/len(df):.4f}')
print(f'사건 E(몸무게 4000g 이상): {E.sum()}개체 / 확률: {E.sum()/len(df):.4f}')
print(f'사건 F(부리 길이 평균 이상): {F.sum()}개체 / 확률: {F.sum()/len(df):.4f}')
```

```python
#실습 (2): 부리길이 데이터를 박스 플롯으로 시각화
import matplotlib.pyplot as plt

penguins=sns.load_dataset('penguins')

plt.figure(figsize=(7, 5))
plt.boxplot(penguins['bill_length_mm'].dropna())
plt.title('bill_length_mm - Box Plot')
plt.xticks([1], ['bill_length_mm'])
plt.show()
```

```python
A=df['species']=='Adelie'
B=df['flipper_length_mm']>=200
```

```python
#실습 (3)

#사건 A와 사건 D의 곱사건(Adelie 종이면서 수컷)
A_and_D=A&D
print(f'A와 D의 곱사건: {A_and_D.sum()}개체 / 확률: {A_and_D.sum()/len(df):.4f}')

#사건 A와 사건 C의 합사건(Adelie 종이거나 Biscoe 섬에 서식)
A_or_C=A|C
print(f'A와 C의 합사건: {A_or_C.sum()}개체 / 확률: {A_or_C.sum()/len(df):.4f}')

#사건 B와 사건 E의 곱사건(지느러미 길이 200mm 이상이면서 몸무게가 4000g 이상)
B_and_E=B&E
print(f'B와 E의 곱사건: {B_and_E.sum()}개체 / 확률: {B_and_E.sum()/len(df):.4f}')

#'사건 D의 여사건'과 '사건 E'의 곱사건(수컷이 아니면서 몸무게가 4000g 이상)
not_D=~D
not_D_and_E=not_D&E
print(f'D의 여사건과 E의 곱사건: {not_D_and_E.sum()}개체 / 확률: {not_D_and_E.sum()/len(df):.4f}')
```
