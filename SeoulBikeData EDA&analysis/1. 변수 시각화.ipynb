```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('SeoulBikeData.csv', encoding='unicode_escape')
df.columns = ['Date', 'Count', 'Hour', 'Temp', 'Humidity', 'Wind', 'Visibility', 
              'Dew', 'Solar', 'Rain', 'Snow', 'Season', 'Holiday', 'Functioning Day']
```

```python
# Date
df['Date'] = pd.to_datetime(df['Date'], format='%d/%m/%Y')

date_summary = pd.DataFrame({
    'Metric': ['Start Date', 'End Date', 'Total Days', 'Unique Days'],
    'Value': [
        df['Date'].min().strftime('%Y-%m-%d'),
        df['Date'].max().strftime('%Y-%m-%d'),
        (df['Date'].max() - df['Date'].min()).days + 1,
        df['Date'].nunique()
    ]
})

date_summary
```

```python
#Count
count_summary = pd.DataFrame({
    'Metric': ['Count', 'Mean', 'Std', 'Min', '25%', '50%', '75%', 'Max'],
    'Value': [
        df['Count'].count(),
        round(df['Count'].mean(), 2),
        round(df['Count'].std(), 2),
        df['Count'].min(),
        df['Count'].quantile(0.25),
        df['Count'].median(),
        df['Count'].quantile(0.75),
        df['Count'].max()
    ]
})

display(count_summary)

plt.figure(figsize=(10, 6))

plt.hist(df['Count'], bins=30, color="#A9D1FC", edgecolor='black')

plt.title('Rented Bike Count Distribution', fontsize=16, fontweight='bold')
plt.xlabel('Rented Bike Count', fontsize=12)
plt.ylabel('Number of Hours', fontsize=12)
plt.grid(axis='y', alpha=0.3, linestyle='--')
plt.savefig('Count.png', dpi=300) # 필요시 주석 해제하여 저장
plt.show()
```

```python
#Hour
hour_summary = pd.DataFrame({
    'Frequency': df['Hour'].value_counts().sort_index(),
    'Percentage (%)': (df['Hour'].value_counts(normalize=True) * 100).round(2)
})
display(hour_summary)

plt.figure(figsize=(12, 6))

ax = hour_summary['Frequency'].plot(kind='bar', color="#A9D1FC", edgecolor='black', width=0.8)

plt.title('Hour Distribution', fontsize=16, fontweight='bold')
plt.xlabel('Hour', fontsize=12)
plt.ylabel('Frequency', fontsize=12)
plt.xticks(rotation=0)
plt.grid(axis='y', alpha=0.3, linestyle='--')

for p in ax.patches:
    ax.annotate(str(int(p.get_height())), (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='center', xytext=(0, 5), textcoords='offset points')

plt.show()
```

```python
#Seasons
seasons_summary = pd.DataFrame({
    'Frequency': df['Season'].value_counts(),
    'Percentage (%)': (df['Season'].value_counts(normalize=True) * 100).round(2)
})
display(seasons_summary)

col = 'Season'
counts = df[col].value_counts()

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# 파이 차트
axes[0].pie(counts, labels=counts.index, autopct='%1.1f%%', startangle=90, colors=sns.color_palette("pastel"))
axes[0].set_title(f'Proportion of {col}')

# 막대 그래프
sns.countplot(data=df, x=col, ax=axes[1], palette="pastel", order=counts.index, saturation=1)
axes[1].set_title(f'Count of {col}')
axes[1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```
